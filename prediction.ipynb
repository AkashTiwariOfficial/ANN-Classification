```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pandas as pd
import numpy as np
```

```python
import pickle

model = load_model('model.h5')

with open("one_hot_code_Geography.pkl", 'rb') as file:
    one_hot_code_Geography=pickle.load(file)

with open("label_encode_gender.pkl", 'rb') as file:
    label_encode_gender=pickle.load(file)

with open("scaler.pkl", 'rb') as file:
    scaler=pickle.load(file)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}


df=pd.DataFrame([input_data])
df
```

```python
code_Geography = one_hot_code_Geography.transform([df['Geography']])
code_Geography.toarray()

new_Geo = pd.DataFrame(code_Geography.toarray(), columns=one_hot_code_Geography.get_feature_names_out(['Geography']))
new_Geo
```

```python
one_hot_code_Geography.get_feature_names_out(['Geography'])
```

```python
df['Gender'] = label_encode_gender.transform(df['Gender'])
df
```

```python
df=pd.concat([df.drop('Geography', axis=1), new_Geo],axis=1)
df
```

```python
input_scaler = scaler.transform(df)
input_scaler
```

```python
prediction = model.predict(input_scaler)
prediction
```

```python
prediction_Probability = prediction[0][0]
prediction_Probability
```

```python
if prediction_Probability>0.5:
    print("will leave the company")
else:
    print("will not leave the company")
```
